```python
import pandas as pd
```

```python
df = pd.read_excel('Dataset/financial_loan_data.xlsx')
df.head()
```

```python
## Fill employee title values
df.fillna({'emp_title': 'Unknown Title'}, inplace=True)
```

#### Feature Engineering

```python
## Extract month data from loan issue date column
df['issue_month_num'] = df['issue_date'].dt.month
df['issue_month_name'] = df['issue_date'].dt.month_name()
```

```python
# PMTD (Previous Month to Date)
previous_month = 11

# MTD (Month to Date)
current_month = 12
```

### KPIs

#### Total Loan Applications

```python
Total_Loan_Applications = df['id'].count()
MTD_Total_Loan_Application = df[df['issue_month_num'] == current_month]['id'].count()
PMTD_Total_Loan_Application = df[df['issue_month_num'] == previous_month]['id'].count()
MoM_Total_Loan_Application = ((MTD_Total_Loan_Application - PMTD_Total_Loan_Application)/PMTD_Total_Loan_Application)*100
```

```python
print(f'Total Loan Applications: {Total_Loan_Applications}')
print(f'MTD Total Loan Applications: {MTD_Total_Loan_Application}')
print(f'PMTD Total Loan Applications: {PMTD_Total_Loan_Application}')
print(f'MoM Total Loan Applications: {MoM_Total_Loan_Application:,.2f}%')
```

#### Total Funded Amount

```python
Total_Funded_Amount = df['loan_amount'].sum()
MTD_Total_Funded_Amount = df[df['issue_month_num'] == current_month]['loan_amount'].sum()
PMTD_Total_Funded_Amount = df[df['issue_month_num'] == previous_month]['loan_amount'].sum()
MoM_Total_Funded_Amount = ((MTD_Total_Funded_Amount - PMTD_Total_Funded_Amount)/PMTD_Total_Funded_Amount)*100
```

```python
print(f'Total Funded Amount: Rs.{Total_Funded_Amount:,.2f}')
print(f'MTD Total Funded Amount: Rs.{MTD_Total_Funded_Amount:,.2f}')
print(f'PMTD Total Funded Amount: Rs.{PMTD_Total_Funded_Amount:,.2f}')
print(f'MoM Total Funded Amount: {MoM_Total_Funded_Amount:,.2f}%')
```

#### Total Received Amount

```python
Total_Received_Amount = df['total_payment'].sum()
MTD_Total_Received_Amount = df[df['issue_month_num'] == current_month]['total_payment'].sum()
PMTD_Total_Received_Amount = df[df['issue_month_num'] == previous_month]['total_payment'].sum()
MoM_Total_Received_Amount = ((MTD_Total_Received_Amount - PMTD_Total_Received_Amount)/PMTD_Total_Received_Amount)*100
```

```python
print(f'Total Received Amount: Rs.{Total_Received_Amount:,.2f}')
print(f'MTD Total Received Amount: Rs.{MTD_Total_Received_Amount:,.2f}')
print(f'PMTD Total Received Amount: Rs.{PMTD_Total_Received_Amount:,.2f}')
print(f'MoM Total Received Amount: {MoM_Total_Received_Amount:,.2f}%')
```

#### Average Interest Rate

```python
Average_Int_Rate = df['int_rate'].mean()*100
MTD_Average_Int_Rate = df[df['issue_month_num'] == current_month]['int_rate'].mean()*100
PMTD_Average_Int_Rate = df[df['issue_month_num'] == previous_month]['int_rate'].mean()*100
MoM_Average_Int_Rate = ((MTD_Average_Int_Rate - PMTD_Average_Int_Rate)/PMTD_Average_Int_Rate)*100
```

```python
print(f'Average Interest Rate: {Average_Int_Rate:,.2f}%')
print(f'MTD Average Interest Rate: {MTD_Average_Int_Rate:,.2f}%')
print(f'PMTD Average Interest Rate: {PMTD_Average_Int_Rate:,.2f}%')
print(f'MoM Average Interest Rate: {MoM_Average_Int_Rate:,.2f}%')
```

#### Average Debt to Income Ratio

```python
Average_Dti_Ratio = df['dti'].mean()*100
MTD_Average_Dti_Ratio = df[df['issue_month_num'] == current_month]['dti'].mean()*100
PMTD_Average_Dti_Ratio = df[df['issue_month_num'] == previous_month]['dti'].mean()*100
MoM_Average_Dti_Ratio = ((MTD_Average_Dti_Ratio - PMTD_Average_Dti_Ratio) / PMTD_Average_Dti_Ratio)*100
```

```python
print(f'Average DTI Ratio: {Average_Dti_Ratio:,.2f}%')
print(f'MTD Average DTI Ratio: {MTD_Average_Dti_Ratio:,.2f}%')
print(f'PMTD Average DTI Ratio: {PMTD_Average_Dti_Ratio:,.2f}%')
print(f'MoM Average DTI Ratio: {MoM_Average_Dti_Ratio:,.2f}%')
```

## Good Loan vs. Bad Loan

```python
## Extract good loan and bad loan data
good_loan = df[df['loan_status'].isin(['Current', 'Fully Paid'])]
bad_loan = df[df['loan_status'] == 'Charged Off']
```

### Good Loan KPIs

```python
Good_Loan_Applications = good_loan['id'].count()
Good_Loan_Application_Pct = (Good_Loan_Applications / Total_Loan_Applications)*100
Good_Loan_Funded_Amount = good_loan['loan_amount'].sum()
Good_Loan_Received_Amount = good_loan['total_payment'].sum()
```

```python
print(f'Good Loan Applications: {Good_Loan_Applications}')
print(f'Good Loan Application Pct: {Good_Loan_Application_Pct:,.2f}%')
print(f'Good Loan Funded Amount: Rs.{Good_Loan_Funded_Amount:,.2f}')
print(f'Good Loan Received Amount: Rs.{Good_Loan_Received_Amount:,.2f}')
```

### Bad Loan KPIs

```python
Bad_Loan_Applications = bad_loan['id'].count()
Bad_Loan_Application_Pct = (Bad_Loan_Applications / Total_Loan_Applications)*100
Bad_Loan_Funded_Amount = bad_loan['loan_amount'].sum()
Bad_Loan_Received_Amount = bad_loan['total_payment'].sum()
```

```python
print(f'Bad Loan Applications: {Bad_Loan_Applications}')
print(f'Bad Loan Application Pct: {Bad_Loan_Application_Pct:,.2f}%')
print(f'Bad Loan Funded Amount: Rs.{Bad_Loan_Funded_Amount:,.2f}')
print(f'Bad Loan Received Amount: Rs.{Bad_Loan_Received_Amount:,.2f}')
```

### Loan Status Grid View

```python
current_df = df[df['issue_month_num'] == current_month]
MTD_FA = current_df.groupby('loan_status')['loan_amount'].sum().reset_index()
MTD_RA = current_df.groupby('loan_status')['total_payment'].sum().reset_index()
```

```python
# Currency Converter
def currencyConverter(dataframe, prefix, *args):
    for col in args:
        dataframe[col] = dataframe[col].map(f'{prefix}''{:,.2f}'.format)
        
# Percentage Converter
def pctConverter(dataframe, suffix, *args):
    for col in args:
        dataframe[col] = (dataframe[col]*100).map('{:,.2f}'f'{suffix}'.format)
```

```python
loan_grid = df.groupby('loan_status').agg(
    Total_Loan_Applications = ('id', 'count'),
    Total_Funded_Amount = ('loan_amount', 'sum'),
    Total_Received_Amount = ('total_payment', 'sum'),
    Average_Int_Rate = ('int_rate', 'mean'),
    Average_Dti_Ratio = ('dti', 'mean')
).reset_index()

loan_grid['MTD_Funded_Amount'] = MTD_FA['loan_amount']
loan_grid['MTD_Received_Amount'] = MTD_RA['total_payment']

currencyConverter(loan_grid, 'Rs.', 'Total_Funded_Amount', 'Total_Received_Amount',
                 'MTD_Funded_Amount', 'MTD_Received_Amount')

pctConverter(loan_grid, '%', 'Average_Int_Rate', 'Average_Dti_Ratio')

loan_grid.columns = loan_grid.columns.str.replace('_', ' ', regex=False) 

loan_grid
```

## Analysis

### Monthly Trend

```python
monthly_trend = df.groupby(['issue_month_num', 'issue_month_name']).agg(
    Total_Loan_Application = ('id', 'count'),
    Total_Funded_Amount = ('loan_amount', 'sum'),
    Total_Received_Amount = ('total_payment', 'sum')
).reset_index().rename(
    columns = {'issue_month_name': 'Month Name'}
)[['Month Name', 'Total_Loan_Application', 'Total_Funded_Amount', 'Total_Received_Amount']]

currencyConverter(monthly_trend, 'Rs.', 'Total_Funded_Amount', 'Total_Received_Amount')

monthly_trend.columns = monthly_trend.columns.str.replace('_', ' ', regex=False)

monthly_trend
```

### Regional Analysis

```python
region_analysis = df.groupby('address_state').agg(
    Total_Loan_Application = ('id', 'count'),
    Total_Funded_Amount = ('loan_amount', 'sum'),
    Total_Received_Amount = ('total_payment', 'sum')
).reset_index().rename(
    columns = {'address_state': 'Region'}
)[['Region', 'Total_Loan_Application', 'Total_Funded_Amount', 'Total_Received_Amount']]

currencyConverter(region_analysis, 'Rs.', 'Total_Funded_Amount', 'Total_Received_Amount')

region_analysis.columns = region_analysis.columns.str.replace('_', ' ', regex=False)

region_analysis
```

### Loan Term Analysis

```python
loan_term = df.groupby('term').agg(
    Total_Loan_Application = ('id', 'count'),
    Total_Funded_Amount = ('loan_amount', 'sum'),
    Total_Received_Amount = ('total_payment', 'sum')
).reset_index().rename(
    columns = {'term': 'Loan Term'}
)[['Loan Term', 'Total_Loan_Application', 'Total_Funded_Amount', 'Total_Received_Amount']]

currencyConverter(loan_term, 'Rs.', 'Total_Funded_Amount', 'Total_Received_Amount')

loan_term.columns = loan_term.columns.str.replace('_', ' ', regex=False)
loan_term
```

### Employee Length Analysis

```python
employee_len_analysis = df.groupby('emp_length').agg(
    Total_Loan_Application = ('id', 'count'),
    Total_Funded_Amount = ('loan_amount', 'sum'),
    Total_Received_Amount = ('total_payment', 'sum')
).reset_index().rename(
    columns = {'emp_length': 'Employee Length'}
)[['Employee Length', 'Total_Loan_Application', 'Total_Funded_Amount', 'Total_Received_Amount']]

currencyConverter(employee_len_analysis, 'Rs.', 'Total_Funded_Amount', 'Total_Received_Amount')

employee_len_analysis.columns = employee_len_analysis.columns.str.replace('_', ' ', regex=False)

employee_len_analysis
```

### Loan Purpose Analysis

```python
loan_purpose = df.groupby('purpose').agg(
    Total_Loan_Application = ('id', 'count'),
    Total_Funded_Amount = ('loan_amount', 'sum'),
    Total_Received_Amount = ('total_payment', 'sum')
).reset_index().rename(
    columns = {'purpose': 'Loan Purpose'}
).sort_values(ascending=False, by='Total_Loan_Application', ignore_index=True)[
    ['Loan Purpose', 'Total_Loan_Application', 'Total_Funded_Amount', 'Total_Received_Amount']
]

currencyConverter(loan_purpose, 'Rs.', 'Total_Funded_Amount', 'Total_Received_Amount')

loan_purpose.columns = loan_purpose.columns.str.replace('_', ' ', regex=False)

loan_purpose
```
