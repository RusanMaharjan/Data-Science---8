```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
server = 'RUSAN'
database = 'BikeStores'

connection_url = f"mssql+pyodbc://@{server}/{database}?driver=ODBC+Driver+17+for+SQL+Server&trusted_connection=yes"
engine = create_engine(connection_url)
```

```python
query = '''
    select
    	CONCAT(sc.first_name, ' ', sc.last_name) as customer_name,
    	COUNT(soi.product_id) as total_products,
    	sum((soi.list_price * soi.quantity) * (1 - soi.discount)) as total_spent
    from sales.customers sc
    join sales.orders so
    on sc.customer_id = so.customer_id
    join sales.order_items soi
    on so.order_id = soi.order_id
    group by CONCAT(sc.first_name, ' ', sc.last_name);
'''
query_df = pd.read_sql(query, engine)
query_df
```

```python
query_sp = '''
    Declare @output_message varchar(100)

    exec usp_findOrderStatus @order_id=1481, @Response_Message = @output_message output;
    
    select @output_message as message;
'''
query_sp_df = pd.read_sql(query_sp, engine)
query_sp_df
```
