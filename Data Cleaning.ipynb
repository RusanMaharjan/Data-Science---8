```python
'''
Data Cleaning
--------------
1. Handle Null Values
    a. Fill Null Values
        - .fillna()

        - If data is continuous
            - Mean
            - Median

            1. If mean and median value of a column is equal or almost equal -> fill mean value
            2. Skewness -> .skew()
                - Right Skewness / Positive Skewness
                    -> data_skew > 0.5 -> fill median value
                    
                - Left Skewness / Negative Skewness
                    -> data_skew < 0 -> fill median value
                
                - Normal Distribution / Bell Curve
                    - 0 - 0.5 -> fill mean value

        - If data is discrete / categorical
            - Mode

        - If data is continuous / discrete / categorical
            - Custom Record
        
    b. Drop Null Values
        - .dropna()

2. Drop Duplicate Values
    - Find duplicate records
        - .duplicated()

    - Drop duplicate records
        - .drop_duplicates()

3. Remove unwated data
    a. Remove whitespaces
        - .str.strip()

    b. Remove unwated values (regular expressions, text, digits)
        - .str.replace()

'''
```

### Removing Outliers uisng IQR Method

```python
hr['Salary'].skew()
```

```python
q1 = hr['Salary'].quantile(0.25)
q3 = hr['Salary'].quantile(0.75)
```

```python
iqr = q3 - q1
```

```python
lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)
```

```python
salary_data = hr[(hr['Salary'] >= lower_bound) & (hr['Salary'] <= upper_bound)]
# salary_data
```

```python
salary_data['Salary'].skew()
```

```python
print(f'Q1: {q1}')
print(f'Q3: {q3}')
print(f'IQR: {iqr}')
print(f'Lower Bound: {lower_bound}')
print(f'Upper Bound: {upper_bound}')
```

#### Remove whitespace

```python
'''
Syntax
-------
col_name.str.strip()
'''
```

```python
def removeWhiteSpaces(dataframe, column_list):
    for col in column_list:
        dataframe[col] = dataframe[col].str.strip()

    print('Whitespace removed.')
```

```python
'''
Pattern of Regular Expressions
-------------------------------------
\d -> Digits
\s -> whitespaces
\w -> words
+ -> One or more
* -> zero or more

sub(pattern, replace_value, data) -> remove data
findall(pattern, data) -> find and extract data
'''
import re # regular expression
```

```python
text = '   Hello   !!   contact us : 54321678  abc@gmail.com   def123@gmail.com 1123345####.'
```

```python
remove_space = re.sub(r'\s+', ' ', text)
remove_space.strip()
```

```python
remove_val = re.sub(r'[!#]', '', remove_space)
remove_val
```

```python
remove_email = re.sub(r'\w+@\w+.\w+', '', remove_space)
remove_email
```

```python
## find email
email = re.findall(r'\w+@\w+.\w+', text)
email
```

```python
'''
DateTime -> YYYY-MM-DD

-> pd.to_datetime(col_name, format/dayfirst)

format
-------
'%d.%m.%y' (default)
'%d-%m-%y'
'mixed'

dayfirst
--------
True
False (default)
'''
```

```python
def convertDateTime(dataframe, *args):
    for col in args:
        dataframe[col] = pd.to_datetime(dataframe[col], format='mixed')
    print('Date converted..')
```
