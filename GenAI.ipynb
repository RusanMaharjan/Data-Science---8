```python
import os
from google import genai
```

```python
gemini_api_key = os.getenv('GEMINI_API_KEY')
```

```python
client = genai.Client()
```

```python
def gemini(prompt):
    interation = client.interactions.create(
        model = "gemini-3.8-flash",
        input = prompt
    )
    return interation.output_text
```

```python
gemini('What is an animal?')
```
