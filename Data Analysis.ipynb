```python
import pandas as pd
```

```python
df = pd.read_csv('cleaned_hr.csv')
df.head()
```

```python
# Find total employees by each department.
dep_emp = df.groupby('Department')['EmpID'].count()
# dep_emp
```

```python
dep_emp = df.groupby('Department')['EmpID'].count().reset_index().rename(
    columns = {'EmpID': 'Total Employees', 'Department': 'Departments'}
).sort_values(ascending=False, by='Total Employees', ignore_index=True)
dep_emp
```

```python
dep_emp = df.groupby('Department')['EmpID'].count().rename_axis('Departments').rename(
    'Total Employees'
).sort_values(ascending=False).to_markdown(stralign='center', numalign='center')
print(dep_emp)
```

```python
def currencyConverter(dataframe:str, prefix:str, *args:str):
    for col in args:
        dataframe[col] = dataframe[col].map(f'{prefix}''{:,.2f}'.format)
```

```python
## Find total employees, total and average salary distributed to employees in each department.
emp_details = df.groupby('Department').agg(
    Total_Employees = ('EmpID', 'count'),
    Total_Salary = ('Salary', 'sum'),
    Average_Salary = ('Salary', 'mean')
).reset_index()
emp_details.columns = emp_details.columns.str.replace('_', ' ', regex=False)
currencyConverter(emp_details, '$', 'Total Salary', 'Average Salary')
# emp_details['Total Salary'] = emp_details['Total Salary'].map('${:,.2f}'.format)
# emp_details['Average Salary'] = emp_details['Average Salary'].map('${:,.2f}'.format)
emp_details
```

```python
## Growth Pct = ((current value - baseline value) / baseline value)*100
# ((45 - 83)/83)*100
```

```python
## Find total number of employees hired in each year.
hire_year_emp = df.groupby('Hire_Year')['EmpID'].count().reset_index().rename(
    columns = {'Hire_Year': 'Hire Year', 'EmpID': 'Total Employees'}
)
hire_year_emp['Hire%'] = (hire_year_emp['Total Employees'].pct_change().fillna(0))*100
hire_year_emp['Hire%'] = hire_year_emp['Hire%'].map('{:,.2f}%'.format)
hire_year_emp
```

```python
## Find total number of employees hired in each month.
emp_hire_month = df.groupby(['Hire_Month', 'Hire_Month_Name'])['EmpID'].count().reset_index().rename(
    columns = {'Hire_Month': 'Hire Month', 'EmpID': 'Total Employees', 'Hire_Month_Name': 'Month Name'}
)[['Month Name', 'Total Employees']]
emp_hire_month
```

```python
## Find total number of employees hired in each day.
# Form category
weekdays = ['Sunday', 'Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday']
df['Hire_Day_Name'] = pd.Categorical(df['Hire_Day_Name'], categories=weekdays)

emp_hire_day = df.groupby(['Hire_Day_Name'], observed=False)['EmpID'].count().reset_index()
emp_hire_day
```
