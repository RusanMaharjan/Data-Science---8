```python
import pandas as pd
```

```python
# 1D -> Series
# 2D -> DataFrame

numbers = pd.Series([1, 2, 3, 4, 5, 6, 7], index=['A', 'B', 'C', 'D', 'E', 'F', 'G'])
numbers
```

```python
product_details = pd.DataFrame([
    ['Mobile', 'Samsung', 2, 50000],
    ['Charger', 'Anker', 5, 3000],
    ['TV', 'MI', 1, 75000],
    ['Bag', 'Bello', 3, 4500]
], columns=['Product Name', 'Product Brand', 'Quantity', 'Price'],
    index = ['A', 'B', 'C', 'D'])
product_details
```

## Loc and ILoc

```python
'''
- Loc -> Location -> Use alphabetical index
- ILoc -> Integer Location -> Use numeric index
'''
```

```python
product_details.iloc[1:3, 1:3]
```

```python
product_details.loc['B':'C', 'Product Brand':'Quantity']
```

```python
product_details.iloc[
    [1, 3], # row
    [0, 2] # column
]
```

```python
product_details.loc[
    ['A', 'D'],
    ['Product Name', 'Quantity', 'Price']
]
```
