```python
'''
Machine Learning
-----------------
1. Types of Machine Learning
    a. Supervised Learning
        -> When labeled data is used.
        -> If our data has both features and target.

        Types of Supervised Learning
        i. Regression
            -> If we have to predict or recommend continuous data.

            Algorithms
            ------------
            a. Linear Regression
            b. Decision Tree Regressor and Random Forest Regressor
            c. KNN (K Nearest Neighbor) -> KNR (K Neighbor Regressor)
            d. SVM (Support Vector Machine) -> SVR (Support Vector Regressor)
            e. XGBoost Regressor

        ii. Classification
            -> If we have to predict or recommend discrete / categorical data.

            Algorithms
            ------------
            a. Logistic Regression
            b. Decision Tree Classifier and Random Forest Classifier
            c. KNN (K Nearest Neighbor) -> KNC (K Neighbor Classifier)
            d. SVM (Support Vector Machine) -> SVC (Support Vector Classifier)
            e. XGBoost Classifier
        
        
    b. Unsupervised Learning
        -> When unlabelled data is used.
        -> If our data only contains features.

        Types of Unsupervised Learning
        i. Clustering

        Algorithm
        -------------
        a. KMeans Clustering
        b. PCA (Principle Component Analysis)
        c. DBSCAN
        d. Agglomerative Clustering
        e. Hierarchical Clustering

2. Process of Implementing Machine Learning
    a. Read Data, Clean Data, Feature Engineering, Data Analysis, Data Visualization
    
    b. Separate Features and Target
        -> All Features must be independent to each other
        -> But Target must be depend on all features.
    
    c. Train Test Split
        -> train_test_split(
            X, Y, test_size=0.3, random_state=42
        )

        Train Test Rule (80/20 Rule)
        - 80% Training Data -> Both Feature and Target Values
        - 20% Testing Data

        70/30, 60/40, 50/50
        
    d. Data Scaling
        -> To stop biasness
        i. Feature Scaling
            a. Standard Scaler
                - If our data is not in range.
                - (Xi - mean) / Sd
                - Best Output Range (-1 to 1)
                - Considerable Output Range (-3 to 3)
                
            b. Min Max Scaler
                - If our data is in range.
                - (Xi - Min_Value) / (Max_Value - Min_Value)
                - Output Range (0 to 1)
            
        ii. Label Encoder
            -> Convert Categorical Data into Numeric Data
            a. Label Encoding
                -> Sort category into ascending order
                -> Then gives sorted category a number starting from 0 to nth value.
                -> Then forms a new column
                
            b. Manual Encoding
            
            c. One Hot Encoding
                -> Sort category into ascending order
                -> Then gives sorted category a number starting from 0 to nth value.
                -> Each cateogry has its own column

    e. Define Model
    f. Fit Training Data into Model
    g. Predict Testing Data
    h. Metrics
    i. Visualization

ML Package
------------
Scikit-Learn (sklearn)
'''
```

## Linear Regression

```python
'''
- Straight Line Concept
- y = mX + C
- y -> Predicted Value
- m -> Slope (B1)
- X -> Data Points
- C -> Intercept / Coefficient (Constant) (B0)

- Euclidean Distance

Metrics
----------
1. MAE (Mean Absolute Error)
    -> Distance between each data points and best fit line
    -> sum of all distance
    -> (x2 - x1) + (y2 - y1)

2. MSE (Mean Squared Error)
    -> Squared Distance between each data points and best fit line
    -> Sum of all squared distance
    -> (x2 - x1)2 + (y2 - y1)2

3. RMSE (Root Mean Squared Error) -> Loss Function
    -> Root(MSE)
    -> Root((x2 - x1)2 + (y2 - y1)2)
    
4. R2 Score (Resudial Score -> E) -> Cost Function -> Accuracy

Rule
----
- If RMSE Increases R2 Score Decreases
- If RMSE Decreases R2 Score Increases -> Correct

Linear Regression Formula
---------------------------
-> If we have multiple features
mlr = B0 + B1.X1 + B2.X2 + ... + Bn.Xn + E

-> If we have single feature
slr -> B0 + B1.X1 + E

Suppose 4 Features -> different 4 Slopes (B1) and Single Intercept (B0)

Types of Linear Regression
---------------------------
1. Polynomial Regression
2. Ridge Regression
3. Lasso Regression
4. Elastic Net Regression
'''
```
