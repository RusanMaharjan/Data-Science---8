```python
import pandas as pd
```

```python
df = pd.read_csv('cleaned_hr.csv')
df.head()
```

```python
def convertDateTime(dataframe, *args):
    for col in args:
        dataframe[col] = pd.to_datetime(dataframe[col], format='mixed')
    print('Date converted..')
```

```python
convertDateTime(df, 'DOB', 'DateofHire', 'DateofTermination', 'LastPerformanceReview_Date')
```

## Extract Date

```python
# dt.year/month/day
df['Hire_Year'] = df['DateofHire'].dt.year
df['Hire_Month'] = df['DateofHire'].dt.month
df['Hire_Day'] = df['DateofHire'].dt.day
df['Hire_Month_Name'] = df['DateofHire'].dt.month_name()
df['Hire_Day_Name'] = df['DateofHire'].dt.day_name()
```

```python
# df[['DateofHire', 'Hire_Year', 'Hire_Month', 'Hire_Day', 'Hire_Month_Name', 'Hire_Day_Name']]
```

## Convert Continuous data to Categorical data

```python
df['Salary'].head()
```

```python
df['Salary'].max() # 250000
df['Salary'].min() # 45000
```

```python
def convertSalary(salary):
    if salary >= 0 and salary <= 60000:
        return 'Low Salary'
    elif salary > 60000 and salary <= 110000:
        return 'Average Salary'
    else:
        return 'High Salary'
```

```python
df['Salar_Label'] = df['Salary'].apply(convertSalary)
df = df.rename(columns = {'Salar_Label': 'Salary_Label'})
```

```python
# df[['Salary', 'Salary_Label']]
```

## Label Encoding

```python
'''
- Convert categorical record to numeric record.
- Types of Encoding
    1. Label Encoder
    2. One-Hot Encoder
    3. Manual Encoding -> Dictionary mapping
'''
```

### Manual Encoding

```python
df['Salary_Label'].unique()
```

```python
sal_dict = {
    'Low Salary': 0, 
    'Average Salary': 1, 
    'High Salary': 2
}
df['sal_id'] = df['Salary_Label'].map(sal_dict)
```

```python
def manualEncoding(dataframe, col_name, *args):
    empty_dict = {}
    for idx, val in enumerate(args):
        empty_dict[val] = idx

    dataframe[f'{col_name}_id'] = dataframe[col_name].map(empty_dict)
```

```python
# df[['Salary', 'Salary_Label', 'sal_id']]
manualEncoding(df, 'Salary_Label', 'Low Salary', 'Average Salary', 'High Salary')
```
